````python
%matplotlib widget
from pathlib import Path
import matplotlib.pyplot as plt
from mpl_pointpicker import PointPicker

image = Path("demo.png")
picker = PointPicker.from_image(image)
picker.ax.figure.set_size_inches(10, 7)
picker.calibrate()
````

## Calibrate, then select

1. Click the x-axis ticks at **0** and **1** (X1, X2).
2. Click the y-axis ticks at **2** and **8** (Y1, Y2).
3. Enter X1=0, X2=1, Y1=2, Y2=8 and press **Apply**.
4. Click the figure to focus it. Add points on either curve with **a**.
5. Press **t** and click a point. Choose **Transition** and label `0,1`, or **Series** and a name such as `series A`. Sigma is optional, in GHz for this graph. Press **Apply**.
6. Use **m** to drag, **d** to delete, **i** to inspect, or **Esc** to return to Add.

Complete these interactions **before running the next cell**. The notebook does not wait for clicks.

````python
picker.view_points()
points = picker.to_dict()
points
````

## Export transitions or all selected coordinates

The transition dictionary omits text groups and unlabeled points. `xy` retains all coordinates. The default transition export contains `(x, y)` pairs; set `include_sigma=True` for triples.

````python
transitions = picker.to_transition_data(include_sigma=True)
print("Transitions:", transitions)
print("All coordinates:", picker.xy)
````

## Save and resume

The YAML file includes points, labels, uncertainties and calibration, but not the image. Provide the image again to resume. Stored coordinates already use final units; loading does not apply calibration twice.

````python
picker.save_yaml("picked-points.yaml")
fig, ax = plt.subplots(figsize=(10, 7))
ax.imshow(plt.imread(image), origin="upper")
restored = PointPicker.load_yaml(ax, "picked-points.yaml")
````

## Work with an existing Matplotlib plot

No image calibration is needed when Matplotlib axes already have the desired coordinates.

````python
fig2, ax2 = plt.subplots()
ax2.plot([0, 1, 2], [1, 4, 2])
plot_picker = PointPicker(ax2)
````

## Release callbacks when finished

Run this final cell only after completing your interactions. Export remains available after disconnecting.

````python
picker.disconnect()
restored.disconnect()
plot_picker.disconnect()
````
