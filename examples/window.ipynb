# Three curves in a separate window

Select the repository's `.venv/bin/python` as the Jupyter kernel. This example uses the native **macOS** Matplotlib window. Run the cells one at a time; restart the kernel first if you previously used an inline or widget backend. The kernel must run locally on your Mac.


```python
%matplotlib osx

from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np

from mpl_pointpicker import PointPicker
```

## Open the window and select points

Use **a** to add, **m** to move, **d** to delete, **t** to label, **i** to inspect, and **n** to start a new x column. The first point fixes x for the current column. Use the toolbar to zoom, then disable zoom before picking. Labels open in a separate Matplotlib form.

Keep `picker` alive. You can leave the window open while running the following cells. Rerunning this cell creates a fresh picker, so save your work first.


```python
x = np.linspace(0, 1, 200)
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(x, 1 + x, label="Curva 1")
ax.plot(x, 2 + 0.3 * np.sin(2 * np.pi * x), label="Curva 2")
ax.plot(x, 3 - 0.5 * x, label="Curva 3")
ax.set(xlabel="x", ylabel="y")
ax.legend()
ax.grid(alpha=0.2)
fig.canvas.manager.set_window_title("PointPicker - tres curvas")

picker = PointPicker(ax, axis_lock=True, use_widgets=False)
plt.show(block=False)
```

## Inspect the selected points

Run this after picking. `xy` includes unlabeled points; `to_dict()` groups labeled points. You can rerun this cell after further edits in the window.


```python
print("All coordinates:")
print(picker.xy)
print("Grouped by label:")
print(picker.to_dict())
```

## Save

Run this when ready. YAML includes unlabeled points too. The printed absolute path shows where to find the file. Rerunning this cell replaces that file. Closing the window keeps the results available in `picker` until the kernel is restarted.


```python
output = Path("jupyter-points.yaml").resolve()
picker.save_yaml(output)
print(f"Saved points to: {output}")
```
