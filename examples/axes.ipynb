# Pick from original Matplotlib data

Keep your original measurement arrays in Matplotlib, zoom into fine details and select several points at the exact same x. No screenshots or image calibration are needed. Install the notebook extra and restart the kernel before running this notebook.

```python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
from mpl_pointpicker import PointPicker

x = np.linspace(0, 1, 2001)
y = np.linspace(2, 8, 801)
ridge = 4 + np.cos(2 * np.pi * x)
signal = np.exp(-(((y[:, None] - ridge[None, :]) / 0.025) ** 2))
fig, ax = plt.subplots(figsize=(5, 4))
mesh = ax.pcolormesh(x, y, signal, shading="auto")
ax.set(xlabel="Control parameter", ylabel="Frequency (GHz)")
picker = PointPicker(ax, axis_lock=True)
```

## Zoom and keep x fixed

1. Use the toolbar zoom rectangle to enlarge a small part of the ridge. Turn zoom **off** before selecting points.
2. Click your first point: its x value becomes the current column.
3. Zoom or pan again if needed, then disable the toolbar tool and add more points. Their x values remain exactly equal.
4. In **m** mode, drag a point: only y changes.
5. Press **n** to begin a new column. To return to a prior column, use **a** and click one of its existing points.
6. Use **t** to assign a transition `0,1`, a series name, and optional sigma.

The QuadMesh retains the original 2001 × 801 samples. View changes only change which region Matplotlib displays. The picker uses the same axes and data coordinates throughout. Complete your interactions before running the next cell.

```python
picker.view_points()
picker.to_dict()
```

## Export and save

All original coordinates remain in `picker.xy`, including unlabeled points. The grouped exports omit unlabeled points. Transition exports also omit text-labeled series.

```python
transitions = picker.to_transition_data(include_sigma=True)
picker.save_yaml("picked-points.yaml")
transitions
```

## Resume on the original data

Recreate the original plot and load the saved point data onto its axes. No image file or calibration is involved.

```python
fig2, ax2 = plt.subplots(figsize=(5, 4))
ax2.pcolormesh(x, y, signal, shading="auto")
restored = PointPicker.load_yaml(ax2, "picked-points.yaml")
```
